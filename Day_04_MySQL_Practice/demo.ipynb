```python
# Null Operator 
- IS NULL: Operator is used to test for empty values (null values).

for example:

SELECT stu_id,stu_name,stu_adress
from student
WHERE stu_name IS NULL;
```

```python
# Update statement

UPDATE student
SET stu_name='Abinash'
where stu_id=1;

- SET statment: it tells database exactly which column to change and what new records to insert.
```
