## Part 2: Variables — Storing Data in Python

Think of a variable as a **labelled box** in your computer's memory. You put a value inside the box, give it a name, and later you can refer to it by that name whenever you need it.

In Python, you do not need to declare what type of data a variable holds before assigning it. Python figures that out automatically. This is called **dynamic typing**.

### Rules for naming variables:
- Must start with a letter or underscore `_` — **not** a number
- Can contain letters, numbers, and underscores
- **Case sensitive** — `name` and `Name` are two different variables
- Cannot be a Python keyword (like `if`, `for`, `while`, `True`, etc.)

### Good naming conventions:
Python developers follow a style called **snake_case** — all lowercase, words separated by underscores. For example: `student_name`, `total_score`, `is_valid`.

**Now let's create some variables. Fill in the values below:**

````python
# Create a variable called 'student_name' and assign your name to it
student_name = "Raj Sharma"

# Create a variable called 'age' and assign your age
age = 22

# Create a variable called 'course' and assign the value 'Data Science'
course = "Data Science"

# Create a variable called 'is_enrolled' and assign True
is_enrolled = True

# Print all four variables
print(student_name)
print(age)
print(course)
print(is_enrolled)
````

### Multiple Assignment

Python allows you to assign values to multiple variables in a single line. This is a neat feature that keeps your code clean. You can also assign the **same value** to multiple variables at once.

**Try both styles below:**

````python
# Assign x=10, y=20, z=30 in a single line
x, y, z = 10, 20, 30

# Print x, y, z
print(x, y, z)

# Assign the value 0 to three variables: a, b, c — all in one line
a = b = c = 0

# Print a, b, c
print(a, b, c)
````

## Part 3: Data Types in Python

Every value in Python has a **type**. The type tells Python what kind of data it is dealing with and what operations are allowed on it. For example, you can multiply two numbers, but you cannot multiply two strings.

Python's core data types are:

| Type | Keyword | Example |
|------|---------|--------|
| Integer | `int` | `42`, `-7`, `0` |
| Floating point | `float` | `3.14`, `-0.5`, `2.0` |
| String | `str` | `'hello'`, `"world"` |
| Boolean | `bool` | `True`, `False` |
| None type | `NoneType` | `None` |

### The `type()` function
Python gives you a built-in function called `type()` that tells you the data type of any value or variable. This is incredibly useful for debugging — when something goes wrong, checking the type is often the first step.

**Explore the types below:**

````python
# Create the following variables and print each one along with its type:

# An integer: total_students = 35
total_students = 35

# A float: gpa = 3.75
gpa = 3.75

# A string: city = 'Kathmandu'
city = 'Kathmandu'

# A boolean: is_passed = True
is_passed = True

# A NoneType: result = None
result = None

# For each variable, print its value AND its type using type()
print(total_students, type(total_students))
print(gpa, type(gpa))
print(city, type(city))
print(is_passed, type(is_passed))
print(result, type(result))
````

### A Closer Look at Strings

Strings are sequences of characters. In Python, you can define a string using either single quotes `'...'` or double quotes `"..."` — both work exactly the same way. Use whichever feels natural, but be consistent.

For **multi-line strings**, Python uses triple quotes: `'''...'''` or `"""..."""`.

Strings also support some useful built-in operations — you can join two strings together (called **concatenation**) using the `+` operator, and you can repeat a string using `*`.

**Observe what happens in the cell below:**

````python

# Create two string variables: first_name and last_name
first_name = "Raj"
last_name = "Sharma"

# Concatenate them to create full_name (with a space in between)
full_name = first_name + " " + last_name

# Print full_name
print(full_name)

# Print the string 'TechAxis ' repeated 3 times using the * operator
print('TechAxis ' * 3)

# Create a multi-line string called 'bio' with at least 2 lines and print it
bio = """Hi, I am a data science student.
I am learning Python syntax and variables today."""
print(bio)
````

## Part 4: Type Conversion (Type Casting)

Sometimes data arrives in the wrong type and you need to convert it. For example, when a user types their age into a form, Python receives it as a **string** — even if the user typed `25`. Before you can do any arithmetic with it, you need to convert it to an integer.

This conversion is called **type casting**. Python provides built-in functions for this:

- `int()` — converts to integer
- `float()` — converts to float
- `str()` — converts to string
- `bool()` — converts to boolean

**Important things to observe:**
- `int('42')` works — `int('hello')` does NOT
- `int(3.9)` gives `3` — it does not round, it **truncates** (cuts off the decimal)
- `bool(0)` is `False`, `bool(1)` is `True`, `bool('')` is `False`, `bool('hello')` is `True`

**Practice type conversion below:**

````python
# Convert the string '100' to an integer and store it in marks
marks = int('100')

# Convert the string '98.6' to a float and store it in temperature
temperature = float('98.6')

# Convert the integer 2024 to a string and store it in year_str
year_str = str(2024)

# Convert the float 7.9 to an integer — observe what happens to the decimal
truncated_int = int(7.9)
print("Truncated 7.9 to:", truncated_int)  # Expected: 3

# Test bool() on: 0, 1, '', 'hello', None
# Print each result and observe the pattern
print("bool(0):", bool(0))
print("bool(1):", bool(1))
print("bool(''):", bool(''))
print("bool('hello'):", bool('hello'))
print("bool(None):", bool(None))
````

## Part 5: Output with `print()`

The `print()` function is how Python communicates back to you. You have already used it, but there is a lot more to it than just `print(something)`.

### Key `print()` features:

**1. Printing multiple items** — you can pass multiple values separated by commas, and `print()` will display them with a space between them by default.

**2. The `sep` parameter** — controls what goes *between* the items. Default is a space `' '`.

**3. The `end` parameter** — controls what goes at the *end* of the print. Default is a newline `'\n'`. If you set `end=''`, the next print continues on the same line.

**4. f-strings (formatted strings)** — the most modern and clean way to embed variables inside strings. You write `f'...'` and put variable names inside `{}`.

**5. The `.format()` method** — an older but still common way to format strings.

**Practise all these below:**

````python
# Print your name, age, and city on the same line using a single print() call
# Use the sep parameter to separate them with ' | '
print(student_name, age, city, sep=' | ')

# Print three numbers 1, 2, 3 on the same line without newlines between them
# Hint: use the end parameter
print(1, end=' ')
print(2, end=' ')
print(3)
````

````python

# Use an f-string to print:
# "My name is [name] and I am [age] years old."
# Use the variables you created earlier
print(f"My name is {student_name} and I am {age} years old.")

# Use an f-string to print a float rounded to 2 decimal places
# Example: price = 1999.5678 → print \"Price: 1999.57\"
price = 1999.5678
print(f"Price: {price:.2f}")
````

````python

# Use .format() to print:
# "Course: Data Science | Batch: 2024"

# your code here
print("Course: {} | Batch: {}".format("Data Science", 2024))
````

## Part 6: Input with `input()`

So far, all the data in our programs has been typed directly into the code. But real programs need to accept data from users at runtime — while the program is running. That is what `input()` does.

### How `input()` works:
- It pauses the program and waits for the user to type something and press Enter
- It displays an optional **prompt message** that you pass as a string argument
- It **always returns a string** — no matter what the user types

This last point is very important and is a common source of bugs for beginners. If you ask a user for their age, you get back a string like `'25'`, not the number `25`. Before doing any calculations, you must cast it to `int` or `float`.

**Let's practice:**

````python
# Ask the user to enter their name using input()
# Store it in a variable called user_name
# Then print: "Hello, [user_name]! Welcome to TechAxis."

# your code here
user_name = input("Enter your name: ")
print(f"Hello, {user_name}! Welcome to TechAxis.")
````

````python

# Ask the user for two numbers
# Convert both to float
# Print their sum, difference, product, and quotient
# Label each result clearly

# your code here
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print(f"Sum: {num1 + num2}")
print(f"Difference: {num1 - num2}")
print(f"Product: {num1 * num2}")

# Guard against division by zero just in case
if num2 != 0:
    print(f"Quotient: {num1 / num2}")
else:
    print("Quotient: Undefined (Cannot divide by zero)")
````

## Part 7: Putting It All Together — Mini Exercise

Now that you have learned about variables, data types, type casting, `print()`, and `input()`, let's combine everything into a small program.

**Task:** Build a simple **Student Profile Generator**.

Your program should:
1. Ask the user for their **name** (string)
2. Ask the user for their **age** (convert to int)
3. Ask the user for their **GPA** (convert to float)
4. Ask the user for their **favourite subject** (string)
5. Print a formatted profile like this:

```
========== Student Profile ==========
Name    : Raj Sharma
Age     : 22
GPA     : 3.85
Subject : Machine Learning
======================================
```

Use f-strings to format the output neatly.

````python

# Student Profile Generator

# Step 1: Ask for name
name = input("Enter student name: ")

# Step 2: Ask for age and convert to int
age_input = int(input("Enter student age: "))

# Step 3: Ask for GPA and convert to float
gpa_input = float(input("Enter student GPA: "))

# Step 4: Ask for favourite subject
subject = input("Enter favourite subject: ")

# Step 5: Print the formatted profile
print("========== Student Profile ==========")
print(f"Name    : {name}")
print(f"Age     : {age_input}")
print(f"GPA     : {gpa_input:.2f}")
print(f"Subject : {subject}")
print("======================================")
````
