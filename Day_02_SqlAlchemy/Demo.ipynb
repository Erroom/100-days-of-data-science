```python
from sqlalchemy import create_engine,text
user='root'
password=''
host='localhost'
port='3306'
database='Practice_SqlAlchemy'
database_url =f"mysql+pymysql://{user}:{password}@{host}:{port}/{database}"
engine=create_engine(database_url,echo=True)
try:
    with engine.connect() as connection:
        result = connection.execute(text("Select 1"))
        print("\nConnection successful! Result of 'SELECT 1':")
except Exception as e:
    print(f"An error occurred: {e}")

```

```python
from sqlalchemy import MetaData, Table, Column, Integer, String
metadata=MetaData()
Students=Table(
    'students',metadata,
    Column('id',Integer,primary_key=True),
    Column('name',String(100)),
    Column('Lastname',String(100))
)
metadata.create_all(engine)
```

```python
#For insert data into table
with engine.connect() as connection:
    ins=Students.insert().values(id=1,name='John',Lastname='Doe')
    connection.execute(ins)
    connection.commit()
```

```python
meta=MetaData()
students=Table(
    'students', meta,
    Column('id',Integer,primary_key=True),
    Column('name',String),
    Column('lastname',String)
)
s=students.select()
conn=engine.connect()
result=conn.execute(s)
for row in result:
    print(row)
```

```python
# Use all uppercase for consistency
s = text("select students.name, students.lastname from students where left(students.name, 1) in ('A', 'J', 'K')")
result = conn.execute(s).fetchall()
```

```python
from sqlalchemy import and_
from sqlalchemy.sql import select
s=select(students).where(
    and_(
        students.c.name.between ('A',"L"),
        students.c.id>2
    )
)
result = conn.execute(s).fetchall()
```
