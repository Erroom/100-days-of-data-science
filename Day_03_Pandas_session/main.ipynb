# Pandas — Session 1: Introduction, Series & DataFrames


NumPy gave us powerful numerical arrays. But real-world data is rarely just numbers in a grid — it has column names, row labels, mixed data types, and missing values. **Pandas** is built exactly for this kind of data.

Pandas is the most widely used data analysis library in Python. It is built on top of NumPy, so everything you learned about NumPy carries forward — Pandas adds a richer, labelled structure on top.

Pandas has two core data structures:

| Structure | What it is | Equivalent to |
|-----------|-----------|---------------|
| `Series` | 1-dimensional labelled array | A single column of data |
| `DataFrame` | 2-dimensional labelled table | A full spreadsheet or SQL table |


````python
## Part 1: Why Pandas?

With a Python list or NumPy array, data has no labels — you access it by position. If you have student data in an array, you have to remember that column 0 is name, column 1 is age, column 2 is GPA. This is fragile and confusing.

Pandas assigns **labels** to both rows and columns. You access data by name, not by position. This makes your code readable, your analysis clear, and your errors much easier to catch.

Beyond labels, Pandas gives you:
- Built-in handling of missing values
- Tools to load data from CSV, Excel, JSON, and databases
- Powerful grouping, filtering, and aggregation
- Integration with Matplotlib for visualisation
- The ability to work with datasets of millions of rows efficiently

By convention, Pandas is always imported as `pd`:
````

## Part 2: Pandas Series

A **Series** is a one-dimensional array with an **index** — a label attached to each value. If you think of a dictionary, a Series is like a more powerful dictionary that supports arithmetic, statistics, and alignment.

```
Index    Value
  0   →   88
  1   →   76
  2   →   91
  3   →   55
```

The index does not have to be numbers — it can be strings (like student names), dates, or anything unique.

```


````python
#Series Operations
import pandas as pd

scores=pd.Series([88,71,91],index=['Aarav','Sita','Binaya'])
bonus=pd.Series([500,100,400],index=['Aarav','Sita','Binaya'])
print(scores)
print("------------------")
print(bonus)
````

Basics of Pandas Series Slicing
Series slicing can be done by using the [:] operator, which allows you to select subset of elements from the series object by specified start and end points.

Below are the syntax's of the slicing a Series −

Series[start:stop:step]: It selects elements from start to end with specified step value.

Series[start:stop]: It selects items from start to stop with step 1.

Series[start:]: It selects items from start to the rest of the object with step 1.

Series[:stop]: It selects the items from the beginning to stop with step 1.

Series[:]: It selects all elements from the series object.

````python
#Example: Slicing the First Three Elements from a Series
s=pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
print(s[:3])
````

## Part 3: Pandas DataFrame

A **DataFrame** is a 2-dimensional table with labelled rows and columns. Think of it as a collection of Series objects that all share the same index — each column is a Series.

This is the primary data structure you will use in all your data science work. Every dataset you load — CSV, Excel, JSON, database — becomes a DataFrame.

```
         name    age    course      gpa
0     Aarav     22     Data Sci    3.8
1      Sita     20     ML          3.6
2    Bikash     23     Data Sci    3.9
```
